### Imports and Hyperparameter Settings

```python
%load_ext autoreload
%autoreload 2

from multimodal_mazes.evolution.genomes.genome_motifs import GenomeMotif as Genome
from multimodal_mazes.evolution.module_banks.motifs import Motif, MotifBank
from multimodal_mazes.evolution.algorithms.genome_EA_motif import GenomeEAMotif

import numpy as np
```

```python
HYPERPARAMETERS = {
    'task': 'maze',
    'n_motif_types': 1,
    'n_inputs': 8,
    'n_outputs': 4,
    'n_motifs': 8,
    'motif_types': ['I1_H1_O1'], # List of motif types to use
    'weight_sharing': True,
    'uniform_weights': True, # For weight sharing
    'connectivity': 'RANDOM', # Options: 'FULLY CONNECTED', 'UNCONNECTED', 'SPARSE', 'RANDOM'
    'connection_density' : {'input_density': 1.0, 'output_density': 1.0}, # For 'RANDOM' connectivity
    'population_size': 1,
    'top_genomes': 1, 
    'mutation_rate': 0.8, 
    'crossover_rate': 0.5,
    'one_to_one': True # Only compatible with UNCONNECTED, SPARSE, and RANDOM if density < 1.0
}
```

```python
types = [
    "I2_O1", 
    "I2_O1.R",
    "I2_O1.T",
    "I2_O1.RT",
    "I1_H1_O1",
    "I1_H1_O1.R",
    "I1_H1_O1.T",
    "I1_H1_O1.RT",
    "I1_O2",
    "I1_O2.R",
    "I1_O2.T",
    "I1_O2.RT"
]

motif_manager = MotifBank(types)
```

## GenomeEA and Genome

### Output Inspection

#### Initialisation

```python
genome_ea = GenomeEAMotif(HYPERPARAMETERS)
genome_ea.generate()
```

```python
genome = genome_ea.genomes[0]

print("Genome ID:", genome.genome_id)

print("Module Rules:")
for rule in genome.mot_rules:
    print(rule)

print("\nInput Connect Rules:")
for rule in genome.conn_in_rules:
    print(rule)

print("\nOutput Connect Rules:")
for rule in genome.conn_out_rules:
    print(rule)

genome.plot_genome()
```

#### Forward Pass

```python
inputs = np.array([1, 0, 0, 0, 0, 0, 0, 0])
print(f"Inputs: {inputs}")

outputs = genome.forward_pass(inputs)
print(f"Outputs: {outputs}")
```

#### Connectivity

```python
genome_ea = GenomeEAMotif(HYPERPARAMETERS)
genome_ea.generate()
genome = genome_ea.genomes[0]
genome.plot_genome()
```

#### Mutation

```python
print("Original Input Connection Rules:")
for rule in genome.conn_in_rules:
    print(rule)

print("\nOriginal Output Connection Rules:")
for rule in genome.conn_out_rules:
    print(rule)

genome.plot_genome()
genome.mutate()
genome.swap_motif()
genome.plot_genome()

print("\nMutated Input Connection Rules:")
for rule in genome.conn_in_rules:
    print(rule)
    
print("\nMutated Output Connection Rules:")
for rule in genome.conn_out_rules:
    print(rule)
```
