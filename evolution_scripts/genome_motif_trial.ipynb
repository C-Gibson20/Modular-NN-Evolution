### Load Results

```python
with open("genome_motif_trial.pkl", "rb") as f:
    trial_objects = pickle.load(f)

ea = trial_objects['ea']
maze = trial_objects['maze']
```

### Fittest Networks

```python
for genome in ea.fittest_networks:
    print(f"Genome ID: {genome.genome_id}, Fitness: {genome.fitness}, Distribution: {genome.mot_dist}")

ea.fittest_networks[0].plot_genome()
```

### Visualising Fittest Network Outputs

```python
path = ea.trial_data[ea.generation-1][ea.fittest_networks[0].genome_id]['paths']
mz = maze[0]

for n in range(5):
    multimodal_mazes.plot_path(path=path[n], mz=mz.mazes[n], mz_goal_loc=mz.goal_locations[n], n_steps=len(path))
```

#### Fitness Over Generations
